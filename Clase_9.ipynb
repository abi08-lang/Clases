 **NumPy**

Un arreglo de NumPy permite operar sobre todos los datos de una vez, sin escribir ciclos.
Celda 2: crear un arreglo y operar con él

```python
# Tiempo de ciclo (minutos) de 6 piezas
tiempos = np.array([4.2, 3.9, 4.5, 4.1, 5.0, 3.8])

print(tiempos)
print(type(tiempos))
print(tiempos.shape)   # (6,) -> 6 elementos en una dimensión

# Convertir todos los tiempos a segundos de un solo paso
segundos = tiempos * 60
print(segundos)

# Estadísticas básicas
print("Promedio:", tiempos.mean())
print("Desviación estándar:", tiempos.std())
print("Mínimo:", tiempos.min())
print("Máximo:", tiempos.max())
```

*Resultado esperado: tiempos * 60 da [252. 234. 270. 246. 300. 228.]; el promedio es 4.25 min y la desviación estándar 0.40.*

**Celda 3: filtrar con una comparación**

```python
# Comparar genera un True/False por cada pieza
print(tiempos > 4.0)

# Quedarse solo con las piezas que tardaron más de 4 minutos
print(tiempos[tiempos > 4.0])
```

**Celda 4: una matriz (filas y columnas)**

```python
# Piezas producidas: 3 máquinas (filas) x 4 semanas (columnas)
produccion = np.array([[120, 135, 128, 140],
                       [ 98, 105, 110, 102],
                       [150, 148, 155, 160]])

print(produccion.shape)          # (3, 4)
print(produccion.sum())          # total general
print(produccion.sum(axis=1))    # total por máquina (suma cada fila)
print(produccion.mean(axis=0))   # promedio por semana (promedia cada columna)
```

*El total general es 1551 piezas y los totales por máquina son [523 415 613]. Para recordar: axis=0 recorre las filas hacia abajo (resultado por columna) y axis=1 recorre las columnas hacia la derecha (resultado por fila).*

**Celda 7: agregar columnas calculadas**

```python
df["Buenas"] = df["Producidas"] - df["Defectuosas"]
df["Pct_defectos"] = (df["Defectuosas"] / df["Producidas"] * 100).round(2)
df["Piezas_por_hora"] = df["Producidas"] / df["Horas"]

df
```

**Celda 8: seleccionar, filtrar y ordenar**

```python
display(df["Producidas"])                 # una columna
display(df[["Dia", "Linea"]])             # varias columnas (doble corchete)
display(df[df["Linea"] == "B"])           # solo las filas de la línea B
display(df[df["Pct_defectos"] > 3])       # filas con más de 3 % de defectos
display(df.sort_values("Pct_defectos", ascending=False).head(3))  # 3 peores
```

El doble signo == compara; un solo = asigna. Las cuatro filas con más de 3 % de defectos son todas de la línea B (días 1, 2, 4 y 5).


**Celda 9: agrupar y resumir**

```python
resumen = df.groupby("Linea")[["Producidas", "Defectuosas"]].sum()
resumen["Pct_defectos"] = (resumen["Defectuosas"] / resumen["Producidas"] * 100).round(2)

print("Línea con mayor % de defectos:", resumen["Pct_defectos"].idxmax())
resumen
```

groupby junta las filas por línea y suma cada columna. El resultado es 2660 piezas y 2.14 % de defectos para la línea A, contra 2395 piezas y 3.97 % para la B.

**Celda 12: histograma (variabilidad de un proceso)**

```python
np.random.seed(42)   # fija la semilla para repetir los mismos números
tiempos_ciclo = np.random.normal(loc=4.2, scale=0.4, size=200)

plt.hist(tiempos_ciclo, bins=15, edgecolor="black")
plt.title("Distribución del tiempo de ciclo")
plt.xlabel("Minutos")
plt.ylabel("Frecuencia")
plt.show()
```
